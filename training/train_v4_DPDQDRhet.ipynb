```python
%%bash

export MPLBACKEND=Agg

cd /content/drive/MyDrive/gpmhc/gpmhc_train

/usr/local/envs/gpmhc/bin/python - <<'PY'

import os
import sys
import json
import torch
import pandas as pd

sys.path.insert(0, os.getcwd())

from torch.utils.data import DataLoader

from gpmhc.learner import json_to_learner
from gpmhc.data import cleanup_schema, tokenize, dataset


PROJECT_ROOT = os.getcwd()


# -----------------------------
# Load model architecture
# -----------------------------
with open(
    "models/baseline_model/json_input.json"
) as f:
    config = json.load(f)

learner = json_to_learner(config)


# -----------------------------
# Load checkpoint weights
# -----------------------------
checkpoint = torch.load(
    "models/baseline_model/model_final.pth",
    map_location="cpu"
)

if "model" in checkpoint:
    learner.model.load_state_dict(checkpoint["model"])
else:
    learner.model.load_state_dict(checkpoint)

learner.model.eval()

print("Weights loaded")


# -----------------------------
# Prepare data
# -----------------------------
df = pd.read_csv(
    "experiments/002_HLAII_heterodimer/dataset/HLAII_test.csv",
    nrows=20
)


df = cleanup_schema(
    df,
    config["dataloader_options"]["csv_to_df"]["schema_options"]
)


x = tokenize(
    df,
    learner.arch.tokenizer
)


y = df["EL"].values.astype("float32")


ds = dataset(x, y)


# -----------------------------
# Use DataLoader
# -----------------------------
loader = DataLoader(
    ds,
    batch_size=5,
    shuffle=False
)


xb, yb = next(iter(loader))


print("\nInput type:", type(xb))
print("Number components:", len(xb))

for i,v in enumerate(xb):
    print(i, type(v), getattr(v, "shape", None))


# -----------------------------
# Forward pass
# -----------------------------
with torch.no_grad():
    out = learner.model(xb)


print("\nOutput:")
print(out)

print("\nLabels:")
print(yb)

PY
```
